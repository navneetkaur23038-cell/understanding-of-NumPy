26. Boolean Masking

```python
marks = np.array([45, 67, 82, 91, 38, 76, 55])

result = marks[marks >= 50]

print("Marks greater than or equal to 50:")
print(result)
```

27. Boolean Masking for High Scores

```python
marks = np.array([45, 67, 82, 91, 38, 76, 55])

high_scores = marks[marks >= 80]

print("High Scores:")
print(high_scores)
```

28. Boolean Masking with Two Conditions

```python
marks = np.array([45, 67, 82, 91, 38, 76, 55])

result = marks[(marks >= 50) & (marks <= 80)]

print("Marks between 50 and 80:")
print(result)
```

29. Broadcasting

```python
array = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

result = array + 5

print("Original Array:")
print(array)

print("\nAfter adding 5:")
print(result)
```

30. Broadcasting with Another Array

```python
array = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

values = np.array([1, 2, 3])

result = array + values

print("Result:")
print(result)
```

36. All Aggregate Functions Together

```python
array = np.array([10, 20, 30, 40, 50])

print("Array:", array)
print("Sum:", np.sum(array))
print("Mean:", np.mean(array))
print("Minimum:", np.min(array))
print("Maximum:", np.max(array))
print("Median:", np.median(array))
```

37. Aggregate Functions on 2D Array

```python
scores = np.array([
    [80, 75, 90],
    [65, 88, 72],
    [95, 91, 89]
])

print("Scores:")
print(scores)

print("\nTotal:", np.sum(scores))
print("Mean:", np.mean(scores))
print("Minimum:", np.min(scores))
print("Maximum:", np.max(scores))
print("Median:", np.median(scores))
```

38. Row-wise Sum

```python
scores = np.array([
    [80, 75, 90],
    [65, 88, 72],
    [95, 91, 89]
])

row_sum = np.sum(scores, axis=1)

print("Row-wise Sum:")
print(row_sum)
```

39. Column-wise Sum

```python
scores = np.array([
    [80, 75, 90],
    [65, 88, 72],
    [95, 91, 89]
])

column_sum = np.sum(scores, axis=0)

print("Column-wise Sum:")
print(column_sum)
```

40. Complete Numerical Data Analysis Example

```python
student_marks = np.array([
    72, 85, 64, 91, 78,
    88, 55, 93, 69, 81
])

print("Student Marks:")
print(student_marks)

print("\n----- Statistical Analysis -----")

print("Total Marks:", np.sum(student_marks))
print("Average Marks:", np.mean(student_marks))
print("Highest Marks:", np.max(student_marks))
print("Lowest Marks:", np.min(student_marks))
print("Median Marks:", np.median(student_marks))

high_scores = student_marks[student_marks >= 80]

print("\nStudents scoring 80 or above:")
print(high_scores)

print(
    "Number of students scoring 80 or above:",
    np.sum(student_marks >= 80)
)
```
